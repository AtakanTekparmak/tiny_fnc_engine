# tiny_fnc_engine Basic Usage

This notebook demonstrates the basic usage of `tiny_fnc_engine`. We will initialize the engine, add some functions, and call them using JSON format.

```python
from tiny_fnc_engine import FunctionCallingEngine
```

```python
# Define some simple functions
def add(a: int, b: int) -> int:
    return a + b

def greet(name: str) -> str:
    return f"Hello, {name}!"
```

```python
# Initialize the engine and register functions
engine = FunctionCallingEngine()
engine.add_functions([add, greet])
```

```python
# Call a single function using a dictionary
single_call = {
    "name": "add",
    "parameters": {"a": 5, "b": 3},
    "returns": [{"name": "result", "type": "int"}]
}

results = engine.parse_and_call_functions(single_call, verbose=True)
print(f"Result: {results[0]}")
```

```python
# Call multiple functions at once
multi_call = [
    {
        "name": "greet",
        "parameters": {"name": "Alice"},
        "returns": [{"name": "greeting", "type": "str"}]
    },
    {
        "name": "add",
        "parameters": {"a": 10, "b": 20},
        "returns": [{"name": "sum", "type": "int"}]
    }
]

results = engine.parse_and_call_functions(multi_call)
for r in results:
    print(r)
```

You can also pass a JSON string directly to `parse_and_call_functions`:

```python
import json

json_call = json.dumps({
    "name": "greet",
    "parameters": {"name": "Bob"},
    "returns": [{"name": "msg", "type": "str"}]
})

results = engine.parse_and_call_functions(json_call)
print(results[0])
```
