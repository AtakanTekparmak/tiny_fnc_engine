# Function Chaining with tiny_fnc_engine

This notebook shows how to chain function calls, where the output of one function is used as input for another.

```python
from tiny_fnc_engine import FunctionCallingEngine
import random
```

```python
def get_random_city() -> str:
    cities = ["New York", "London", "Tokyo", "Paris", "Sydney"]
    return random.choice(cities)

def get_weather_forecast(city: str) -> dict:
    conditions = ["Sunny", "Rainy", "Cloudy", "Snowy"]
    temp = random.randint(0, 35)
    return {"city": city, "condition": random.choice(conditions), "temperature": temp}
```

```python
engine = FunctionCallingEngine()
engine.add_functions([get_random_city, get_weather_forecast])
```

```python
# Chain the functions: get_random_city -> get_weather_forecast
chain_call = [
    {
        "name": "get_random_city",
        "parameters": {},
        "returns": [{"name": "random_city", "type": "str"}]
    },
    {
        "name": "get_weather_forecast",
        "parameters": {"city": "random_city"},  # References the output of the previous function
        "returns": [{"name": "forecast", "type": "dict"}]
    }
]

results = engine.parse_and_call_functions(chain_call, verbose=True)

print(f"Selected City: {results[0]}")
print(f"Weather Forecast: {results[1]}")
```

### Note on Resetting Sessions
If you plan to run chained calls multiple times with different parameters, you can clear the stored outputs using `engine.reset_session()`.

```python
engine.reset_session()
print("Session cleared. Outputs dict is now empty.")
```
